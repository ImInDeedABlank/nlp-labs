# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

docs = [
    "This is the first document.",
    "This document is the second document.",
    "And this is the third one.",
    "Is this the first document?",
]

tfidf = TfidfVectorizer()
matrix = tfidf.fit_transform(docs)

labels = [f"d{i+1}" for i in range(len(docs))]
sim = pd.DataFrame(cosine_similarity(matrix), index=labels, columns=labels)
sim.round(3)
````

````python
# Same numbers, written out, plus the most similar pair.
print("cosine similarity matrix")
print(sim.round(3).to_string(), "\n")

# upper triangle only, to skip the diagonal (always 1.0) and the mirror pairs
pairs = [(labels[i], labels[j], sim.iloc[i, j])
         for i in range(len(docs)) for j in range(i + 1, len(docs))]

print("pairs, most similar first")
for a, b, s in sorted(pairs, key=lambda t: -t[2]):
    print(f"{a} - {b}: {s:.3f}   |  {docs[labels.index(a)][:38]!r} vs {docs[labels.index(b)][:38]!r}")

# d1 and d4 score exactly 1.000: they are the same five words in a different
# order (this / is / the / first / document). TF-IDF is a bag of words, so it
# cannot see order at all - "This is the first document." and "Is this the
# first document?" collapse to one and the same vector. That is the limitation
# to remember about this representation.
# d2 sits at 0.647 against both: it shares this/is/the/document but adds
# "second" and repeats "document", so it points in a similar but not identical
# direction.
# d3 is the furthest from everything (0.225 - 0.308). It shares only
# this/is/the with the others; "and", "third" and "one" appear nowhere else.
````

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

sentences = [
    "data science is one of the most important fields of science",
    "this is one of the best data science courses",
    "data scientists analyze data",
]

tfidf = TfidfVectorizer()
matrix = tfidf.fit_transform(sentences)

tfidf_df = pd.DataFrame(
    matrix.toarray(),
    columns=tfidf.get_feature_names_out(),
    index=[f"s{i+1}" for i in range(len(sentences))],
)
tfidf_df.round(3)
````

````python
# The important words are the ones with the highest weight in each sentence.
print("top TF-IDF terms per sentence")
for i, row in tfidf_df.iterrows():
    top = row.sort_values(ascending=False).head(4)
    top = [f"{w} ({v:.3f})" for w, v in top.items() if v > 0]
    print(f"  {i}: {', '.join(top)}")
print()

# Why the ranking comes out this way: TF is local to the paragraph, IDF is
# global to the corpus.
#   "science" (0.488) appears twice in the 11-token s1 and in no other
#     sentence, so it has high TF and high IDF at the same time.
#   "of" ties it at 0.488 - strong in s1, absent from the other two.
#   "data" tops s3 (0.641) even though it is in all three sentences, because
#     TF dominates there: 2 of the 4 tokens in s3 are "data".
#   "the" / "is" / "of" do appear in the table, but a word present in every
#     sentence has its IDF at the floor and cannot reach the top of a row on
#     frequency alone.
#
# The floor is not 0. scikit-learn's default is
#   idf = ln((1 + n) / (1 + df)) + 1
# so a word appearing in all 3 documents scores exactly ln(4/4) + 1 = 1.0, and
# one appearing in 2 of 3 scores ln(4/3) + 1 = 1.288. The +1 is there to stop
# IDF from ever reaching zero.
idf_order = tfidf.idf_.argsort()
print("lowest IDF terms (the most common across the 3 sentences):")
for word, value in zip(tfidf.get_feature_names_out()[idf_order][:6],
                       tfidf.idf_[idf_order][:6]):
    print(f"   {word:<12} idf={value:.3f}")
````

````python
import re
import os
import pandas as pd
import gensim
from nltk.tokenize import word_tokenize

# Simpsons script lines, shipped next to this notebook.
csv_path = "dataset/simpsons_script_lines.csv"
if not os.path.exists(csv_path):                      # fallback if cwd differs
    csv_path = os.path.join(os.path.dirname(os.path.abspath(".")), "dataset/simpsons_script_lines.csv")

simpsons = pd.read_csv(csv_path, low_memory=False)
print("shape:", simpsons.shape)
print("columns:", list(simpsons.columns))
simpsons[["raw_character_text", "spoken_words"]].head()
````

````python
import nltk
for _pkg in ("punkt", "punkt_tab"):
    nltk.download(_pkg, quiet=True)

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", "", text)
    text = re.sub(r"[)(,”“.’$-]", "", text)
    return text

lines = simpsons["spoken_words"].dropna().astype(str).map(clean_text)

# Word2Vec wants a list of token lists, one list per line of dialogue.
tokens = [word_tokenize(line) for line in lines]
tokens = [t for t in tokens if t]

print("lines:", len(tokens))
print(tokens[0])

Skip_gram_model = gensim.models.Word2Vec(
    tokens, min_count=1, vector_size=100, window=5, sg=1
)

print("vocabulary size:", len(Skip_gram_model.wv))
````

### Task 4

Use: wv.most_similar() method to :

1.	Find the words similar to “homer”.

2.	Find the words similar to “marge”.

3. Find the words similar to “bart”



````python
# wv.most_similar() returns the words whose vectors sit closest to the target
# in cosine terms. sg=1 (skip-gram) is what makes this work for character
# names: a name learned from the words that surround it ends up near the
# other names that share those surroundings.
for name in ["homer", "marge", "bart"]:
    print(f"words most similar to '{name}'")
    for word, score in Skip_gram_model.wv.most_similar(name, topn=8):
        print(f"   {word:<18} {score:.3f}")
    print()
````

### Task 5

Use the wv.doesnt_match() method to :

1.	Find which of 'jimbo', 'milhouse’, and 'kearney’ does not belong to the list.

3.	Find the odd one among "nelson", "bart", and "milhouse".

4.	Find the odd one among ‘homer', 'patty', and ‘selma'.

*Hint: You need to pass the strings as List*

````python
# doesnt_match takes a list and returns the element whose vector is least
# similar to the average of the rest - the odd one out.
groups = [
    ["jimbo", "milhouse", "kearney"],
    ["nelson", "bart", "milhouse"],
    ["homer", "patty", "selma"],
]

for group in groups:
    odd = Skip_gram_model.wv.doesnt_match(group)
    print(f"{str(group):<45} -> odd one out: {odd}")

# Sanity check on the first group: the pairwise similarities.
from itertools import combinations
print()
for a, b in combinations(groups[0], 2):
    print(f"  similarity({a}, {b}) = {Skip_gram_model.wv.similarity(a, b):.3f}")
````
