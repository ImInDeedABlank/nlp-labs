# Disneyland Reviews Dataset
In this lab we'll look into Disneyland Reviews dataset.  
*Link: https://www.kaggle.com/datasets/arushchillar/disneyland-reviews*

The dataset includes 42,000 reviews of 3 Disneyland branches - Paris, California and Hong Kong, posted by visitors on Trip Advisor.

Column Description:

- Review_ID: unique id given to each review
- Rating: ranging from 1 (unsatisfied) to 5 (satisfied)
- Year_Month: when the reviewer visited the theme park
- Reviewer_Location: country of origin of visitor
- Review_Text: comments made by visitor
- Disneyland_Branch: location of Disneyland Park

```python
%pip install -q kagglehub
# Load the dataset
import os
import kagglehub
import pandas as pd

# DisneylandReviews.csv ships in this Kaggle dataset; kagglehub downloads it
# once and returns the folder it was cached in.
DISNEY_DIR = kagglehub.dataset_download("arushchillar/disneyland-reviews")

data = pd.read_csv(os.path.join(DISNEY_DIR, 'DisneylandReviews.csv'), encoding='latin-1')
data
```

```python
#Check for missing values
data.isna().sum()
```

```python
# Drop rows with missing values in the relevant columns
data = data.dropna(subset=['Review_Text', 'Rating'])
```

```python
# Map star ratings to sentiments
data['Sentiment'] = data['Rating'].apply(lambda rating: 'positive' if rating > 3 else ('negative' if rating < 3 else 'neutral'))
data.head()
```

```python
# Import necessary libraries
import re
import nltk

def preprocess_text(text):

    # 1. Convert text to lowercase
    text = text.lower()

    # 2. Remove punctuation and special characters
    text = re.sub(r'[^a-z\s]', '', text)

    # 3. Tokenize the text
    words = text.split()

    return ' '.join(words)

# Apply preprocessing to the review text
data['Review_Text'] = data['Review_Text'].apply(preprocess_text)

# Display cleaned reviews
data[['Review_Text', 'Sentiment']].head()
```

```python
# Import necessary libraries
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

# Split the dataset into training and testing sets
train_data, test_data, train_labels, test_labels = train_test_split(data['Review_Text'], data['Sentiment'], test_size=0.2, random_state=42)


# TF-IDF Vectorization
tfidf_vectorizer = TfidfVectorizer(max_features=1000)  # You can adjust max_features based on your dataset size - limit vocabulary to the 1000 most informative tokens
train_vectors = tfidf_vectorizer.fit_transform(train_data)
test_vectors = tfidf_vectorizer.transform(test_data)
```

```python
print(f"train_data size: {train_data.shape}")
print(f"test_data size: {test_data.shape}")
```

```python
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report

# Train the Support Vector Machine (SVM) classifier
svm_classifier = SVC(kernel='linear')
svm_classifier.fit(train_vectors, train_labels)

# Predictions on the test set
predictions = svm_classifier.predict(test_vectors)

# Evaluate the model
accuracy = accuracy_score(test_labels, predictions)
print(f"Accuracy: {accuracy:.2f}")

# Display classification report
print("Classification Report:")
print(classification_report(test_labels, predictions))
```

# Use Case:  Amazon reviews of unlocked phone analysis

The objective of this use case is to conduct sentiment analysis on Amazon reviews of unlocked phones, categorizing reviews into three classes: positive, negative, and neutral. The goal is to gain a comprehensive understanding of customer opinions and sentiments regarding various unlocked phone models.

- Dataset link: https://www.kaggle.com/datasets/PromptCloudHQ/amazon-reviews-unlocked-mobile-phones

Task1: Load the data and do 5 different preprocessing steps on it

Task2: Use a proper train and test split

Task3: Do Feature Extraction Using TF-IDF

Task4: Train a Naive Bayes Classifier

Task5: Evaluate the model on the Test Set

Task6: Print the Confusion Matrix

```python
import os
import re
import kagglehub
import numpy as np
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

# Use case: sentiment analysis on Amazon reviews of unlocked mobile phones.
# https://www.kaggle.com/datasets/PromptCloudHQ/amazon-reviews-unlocked-mobile-phones
DATA_DIR = kagglehub.dataset_download("PromptCloudHQ/amazon-reviews-unlocked-mobile-phones")
df = pd.read_csv(os.path.join(DATA_DIR, "Amazon_Unlocked_Mobile.csv"))

print("shape:", df.shape)
print("columns:", list(df.columns))
df.head(3)
```

```python
import nltk
for _pkg in ("punkt", "punkt_tab", "stopwords", "wordnet", "omw-1.4"):
    nltk.download(_pkg, quiet=True)

# ---- Task 1: five preprocessing steps ----------------------------------
# Step 0 (cleanup that makes the other steps possible): drop rows with no
# review text and coerce Rating to a number, since a handful of rows hold junk.
df = df.dropna(subset=["Reviews", "Rating"]).copy()
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
df = df.dropna(subset=["Rating"])

# The file has ~413k reviews. Tokenising + lemmatising all of them takes
# several minutes in NLTK, so we work on a reproducible 30k sample. Every later
# step is unchanged, just smaller.
df = df.sample(n=30_000, random_state=42).reset_index(drop=True)

# 3-class target, same mapping the Disneyland example used.
df["Sentiment"] = df["Rating"].apply(
    lambda r: "positive" if r > 3 else ("negative" if r < 3 else "neutral")
)
print(df["Sentiment"].value_counts())
print()
print(df.loc[0, "Reviews"][:300])
```

```python
STOPWORDS = set(stopwords.words("english"))
LEMMATIZER = WordNetLemmatizer()

def preprocess_text(text):
    """Five steps: lowercase, strip markup/punctuation/digits, tokenize,
    drop stopwords, lemmatize."""

    # 1. lowercase - "Phone" and "phone" are the same word
    text = str(text).lower()

    # 2. remove markup, URLs, punctuation and digits with one regex.
    #    Everything that is not a letter or whitespace becomes a space, so
    #    "don't" -> "don t" rather than "dont".
    text = re.sub(r"<[^>]+>", " ", text)        # html tags such as <br />
    text = re.sub(r"https?://\S+|www\.\S+", " ", text)
    text = re.sub(r"[^a-z\s]", " ", text)

    # 3. tokenize
    words = word_tokenize(text)

    # 4. remove stopwords ("the", "and", "is", ...) - they appear in almost
    #    every review so they carry no signal about sentiment
    words = [w for w in words if w not in STOPWORDS and len(w) > 1]

    # 5. lemmatize - collapse inflected forms onto the dictionary form.
    #    pos="v" first so that "used"/"using" agree; default pos is noun.
    words = [LEMMATIZER.lemmatize(w, pos="v") for w in words]

    return " ".join(words)

df["clean_review"] = df["Reviews"].apply(preprocess_text)

print("before:", df.loc[0, "Reviews"][:200])
print()
print("after :", df.loc[0, "clean_review"][:200])
```

```python
from sklearn.model_selection import train_test_split

# ---- Task 2: train / test split ---------------------------------------
# stratify keeps the class mix identical in both halves. It matters here
# because "neutral" is a small minority class and a plain random split can
# leave the test set with very few of them.
X_train, X_test, y_train, y_test = train_test_split(
    df["clean_review"],
    df["Sentiment"],
    test_size=0.2,
    random_state=42,
    stratify=df["Sentiment"],
)

print("train:", X_train.shape, " test:", X_test.shape)
print("train labels:", y_train.value_counts().to_dict())
print("test  labels:", y_test.value_counts().to_dict())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# ---- Task 3: TF-IDF feature extraction --------------------------------
# Fit on the training text only. Fitting on train+test would leak the test
# vocabulary and the test IDF values into the model and inflate the score.
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1, 2), min_df=3)

X_train_tfidf = tfidf.fit_transform(X_train)     # learn vocabulary + idf
X_test_tfidf = tfidf.transform(X_test)           # apply the same mapping

print("train matrix:", X_train_tfidf.shape)
print("test  matrix:", X_test_tfidf.shape)
print("non-zero entries:", X_train_tfidf.nnz)
print()
print("a few learned features:", tfidf.get_feature_names_out()[:20])
```

```python
from sklearn.naive_bayes import MultinomialNB

# ---- Task 4: Naive Bayes ----------------------------------------------
# MultinomialNB is the right variant for count/TF-IDF features: it models how
# often each word occurs, and treats every feature as independent given the
# class (the "naive" part). alpha is Laplace smoothing; without it any word
# unseen in a class would zero out the whole posterior.
nb = MultinomialNB(alpha=1.0)
nb.fit(X_train_tfidf, y_train)

print("classes:", nb.classes_)
print("log priors:", dict(zip(nb.classes_, nb.class_log_prior_.round(3))))
```

```python
from sklearn.metrics import accuracy_score, classification_report

# ---- Task 5: evaluate on the test set ---------------------------------
predictions = nb.predict(X_test_tfidf)

print(f"Accuracy: {accuracy_score(y_test, predictions):.4f}")
print()
print(classification_report(y_test, predictions, digits=3, zero_division=0))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix
import matplotlib.pyplot as plt

%matplotlib inline

# ---- Task 6: confusion matrix -----------------------------------------
cm = confusion_matrix(y_test, predictions, labels=nb.classes_)

print("rows = true label, columns = predicted label")
print(pd.DataFrame(cm, index=nb.classes_, columns=nb.classes_))
print()

fig, ax = plt.subplots(figsize=(5, 4))
ConfusionMatrixDisplay(cm, display_labels=nb.classes_).plot(ax=ax, colorbar=False, cmap="Blues")
plt.tight_layout()
plt.show()

# What the matrix says: the model is accurate on the two big classes and
# almost never predicts "neutral" - neutral reviews read like positive or
# negative ones, and 3-star reviews are genuinely mixed text.
```

**Reading the results**

Accuracy looks respectable, but the per-class numbers tell the real story: recall on `neutral` is the column to look at. Reviews rated 3 stars are not "in between" in language - plenty of them are enthusiastic text with one complaint - so the bag-of-words model has little to separate them from 4- and 5-star reviews. The usual fixes are to merge neutral into a binary problem, or to weight the classes (`class_weight="balanced"` on the Naive Bayes, or resampling the training set).
